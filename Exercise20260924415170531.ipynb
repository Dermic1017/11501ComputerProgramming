## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
因為 Python 的 int 採用 Arbitrary Precision（任意精度）設計，可以依需求動態增加儲存空間。
2. Python 的 `int` 是否具有固定的位元數限制？
沒有固定的位元數限制。
3. 什麼是 Arbitrary Precision（任意精度）？
任意精度代表整數的位數不受固定限制，當數字變大時，Python 會自動配置更多記憶體來儲存。
4. CPython 使用什麼物件來實作 `int`？
PyLongObject
5. Python 整數大小真正的限制是什麼？
受到電腦可用記憶體大小限制，而非固定的位元數限制。

1. 為什麼結果與數學上的答案不同？
因為電腦使用二進位浮點數表示小數，部分小數無法被精確表示，因此會產生誤差。

不能。
3. 什麼是浮點數的精度（precision）？
能夠正確表示的有效數字位數。
4. 什麼是浮點數的範圍（range）？
可以表示的最大值與最小值範圍。
5. 精度與範圍有何不同？
精度代表數值的細緻程度；範圍代表可表示數值的大小區間。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？
浮點數（Floating Point Numbers）。
2. 什麼是 FP64？
64 位元浮點數（Double Precision），精度最高，但記憶體使用量較大。
3. 什麼是 FP32？
32 位元浮點數（Single Precision），目前深度學習常用格式。
4. 什麼是 FP16？
16 位元浮點數，精度較低但速度較快，可節省記憶體。
5. 什麼是 BF16？
Brain Floating Point 16，保留較大的指數範圍，常用於大型 AI 模型訓練。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？
因為權重通常需要表示小數，且浮點數運算速度快、記憶體使用量較小，更適合 AI 訓練與推論。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？
Unlimited Precision（無限制精度）表示整數的位數沒有固定上限，Python 可以依照需要使用更多記憶體來儲存更大的整數。
2. 它是否代表可以儲存無限大的整數？
理論上不是無限大，而是只要電腦記憶體足夠，就可以儲存非常大的整數。
3. CPython 如何實現這項特性？
CPython 使用 PyLongObject 實作整數，並以動態配置記憶體的方式儲存數值，因此可以隨著整數位數增加而擴大儲存空間。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
因為 Python 的 int 採用 Arbitrary Precision 設計，會根據數值大小自動使用更多記憶體。
2. 為什麼 `float` 卻存在範圍限制？
因為 float 採用 IEEE 754 浮點數格式，使用固定的 64 位元儲存，因此可表示的範圍有限。
3. 為什麼 `float` 可能產生誤差？
因為許多十進位小數無法被二進位完全表示，因此在儲存時會產生近似值，造成計算誤差。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 的整數強調正確性與精確度，而 AI 常用的 FP16、BF16 則強調運算速度與記憶體效率。
2. 各自的優點與缺點。

Python int：
優點：精確、不會因位數不足而溢位。
缺點：佔用較多記憶體、運算較慢。

FP16 / BF16：
優點：記憶體需求小、運算速度快。
缺點：精度較低，可能產生誤差。

3. 為什麼大型語言模型通常選擇較低精度的表示方式？
因為大型模型擁有數十億甚至數千億個參數，使用低精度格式可以大幅降低記憶體需求並提升訓練與推論速度。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
記憶體消耗大幅增加、訓練速度變慢、GPU 可容納的模型大小減少，導致成本提高。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

Python 可以正常運作 x = 10**1000，因為 Python 的 int 採用 Arbitrary Precision（任意精度）設計。當整數變得非常大時，CPython 會透過 PyLongObject 動態配置更多記憶體來儲存數值，因此不受固定 32 位元或 64 位元整數限制。
 
然而大型 AI 模型通常使用 FP32、FP16 或 BF16，而不是任意精度整數或 FP64，原因如下：
 
1. 記憶體使用量
FP16 與 BF16 僅需 16 位元，而 FP32 需要 32 位元，能大幅減少模型儲存空間。若使用 FP64，記憶體需求會增加許多；若使用任意精度整數，成本更高。
 
2. 計算速度
現代 GPU 對 FP16 與 BF16 有專門硬體加速，因此運算速度遠快於高精度格式。
 
3. 數值精度
AI 訓練不需要每次計算都完全精確。雖然 FP16、BF16 會有些微誤差，但通常不會影響模型最終效果。
 
4. GPU 運算效率
較低精度格式可以讓同一張 GPU 同時處理更多資料，提高訓練與推論效率，因此大型 AI 模型普遍採用 FP16、BF16 或 FP32。
 
總結來說，Python 的 int 設計追求數值正確性，而 AI 模型更重視記憶體效率與運算速度，所以選擇較低精度的浮點數格式。
