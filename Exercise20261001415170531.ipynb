### 1. 0.1 的二進位表示法
 
0.1 轉換成二進位後會變成：
 
0.0001100110011001100110011...
 
可以看到 0011 會無限循環，因此無法被有限位元精確表示。
 
Python 使用 IEEE 754 浮點數格式儲存數值，因此只能儲存近似值，而不是完全精確的 0.1。

### 2. Python 如何解決無限多位問題
 
Python 採用 IEEE 754 浮點數標準來儲存浮點數。
 
由於記憶體空間有限，無法儲存無限位數，因此會使用最接近的近似值來表示。
 
在進行捨入時，IEEE 754 使用「四捨六入五成雙」（Round Half to Even）的方法，以減少大量計算時累積的誤差。

### 3. 輸出結果說明
 
sys.float_info 會顯示 Python 浮點數的相關資訊。
 
重要欄位：
 
- max：可表示的最大浮點數
- min：可表示的最小正浮點數
- dig：有效十進位數字數量
- epsilon：最小可分辨誤差
 
Python 的 float 使用 IEEE 754 雙精度浮點數格式儲存資料。

````python
import sys
print(sys.float_info)
````

### 4. 原因說明
 
程式輸出結果為：
 
0.30000000000000004
 
因為 0.1 與 0.2 在二進位中都無法被精確表示。
 
Python 儲存的是近似值，因此相加後會產生極小的誤差，所以結果不是完全的 0.3。

````python
print(0.1 + 0.2)
````

### 5. 我的方法
 
若需要高精度計算，例如金融計算或科學計算，可以使用 Decimal 模組進行計算。
 
例如：
 
Decimal("0.1") + Decimal("0.2")
 
結果會得到精確的 0.3。
 
因此一般浮點數計算不一定適合金融計算，因為誤差可能會不斷累積。
