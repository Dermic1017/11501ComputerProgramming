為自己學Python

安裝 Python 的困難度不高，基本上就是到 Python 的官方網站下載安裝程式並照著指示做就行了，目前主流的作業系統，例如 Windows、macOS 以及 Linux 都有相對應的版本。是說，如果各位使用的是 macOS 跟 Linux 作業系統，可能會發現原本電腦裡就已經直接內建 Python 了，這樣還需要另外安裝嗎？作業系統內建安裝的 Python 通常會比現行版本稍微舊一點，如果單純就以體驗或寫一些簡單的程式的話是夠用的。作業系統之所以會內建安裝 Python 是有原因的，因為有些系統程式本身就是用 Python 開發的。Python 的版本從早期的 2.x 版到現在的 3.x 版，除了正宗的 CPython 之外，還有其他分支的實作品版本（像是 Jython、IronPython、PyPy 等），如果不想把自己的電腦環境弄的亂七八糟，或是因為專案需要常常安裝不同版本的 Python，通常會建議另外使用版本管理工具來建構隔離的開發環境，這裡將使用 pyenv 做為範例。

可以把 pyenv 想像成一個工具箱，在這個工具箱裡會可以放很多種工具，但這個工具箱裡如果沒有任何工具它就只是個空的盒子而已。pyenv 不只是個工具箱，還是有分隔板的工具箱，不同版本的 Python 會被安裝在不同的隔板裡，安裝套件的時候也會安裝在各自的隔板裡不會互相干擾，透過 pyenv 就能做到簡單的切換 Python 的環境及版本。

提醒一下，如果你的作業系統（例如 macOS）原本就有預先安裝 Python 的話，即使透過 pyenv 安裝了其他版本的 Python，原本系統內建的 Python 也建議不要移除，它存在作業系統裡一定有原因的，你不會知道系統裡有哪些程式正在使用它。

pyenv 的主要功能包括：

- 安裝多個 Python 版本
- 切換全域（global）Python 版本
- 為不同專案指定 Python 版本
- 管理 CPython、PyPy 等不同 Python 發行版本

1. pyenv 的主要用途是什麼？

   pyenv是一個Python版本管理工具。主要用途是在同一台電腦或同一個開發環境中，安裝並切換不同版本的Python，而不必移除原本的版本。

2. 為什麼有些開發者會在同一台電腦安裝多個 Python 版本？

   有些開發者在同一台電腦安裝多個 Python 版本，是因為不同專案可能需要不同版本的 Python。這就像有些文件只能用特定版本的軟體開啟，程式和套件也可能只支援某些 Python 版本。

3. 請說明下列指令的用途：

   (a) `pyenv versions`

   顯示目前透過 pyenv 安裝的所有 Python 版本，並以 * 標示目前正在使用的版本。

   (b) `pyenv install 3.12.7`

   下載並安裝 Python 3.12.7，使電腦可以使用這個 Python 版本。這個指令只負責安裝，不代表安裝後一定會立即切換到該版本。

   (c) `pyenv shell 3.12.7`

   將目前終端機工作階段暫時切換成 Python 3.12.7。設定只對目前開啟的終端機有效，關閉終端機後通常就會失效。

   (d) `pyenv global 3.12.7`

   將 Python 3.12.7 設定為目前使用者的預設 Python 版本。當專案沒有另外設定 local 版本時，通常會使用這個全域版本。

   (e) `pyenv local 3.12.7`

   將目前資料夾及其子資料夾的 Python 版本設定為 3.12.7。執行後通常會在目前資料夾建立一個 .python-version 檔案，用來記錄該專案應使用的 Python 版本。

4. `shell`、`local` 與 `global` 三種設定方式有何差異？

shell：只影響目前的終端機工作階段，關閉終端機後通常失效。
local：影響目前專案資料夾及其子資料夾，通常記錄在 .python-version 中。
global：設定目前使用者平常預設使用的 Python 版本。
shell 的優先權最高，接著是 local，然後才是 global

5. pyenv 與 venv 有何不同？

   pyenv 主要負責管理和切換不同的 Python 版本，例如 Python 3.10、3.12 和 3.13。

   venv 主要負責為不同專案建立獨立的套件環境，避免不同專案安裝的 numpy、pandas 或其他套件互相衝突。venv 建立的環境是以目前已存在的 Python 版本為基礎。

6. Git、pyenv 與 venv 分別負責管理什麼？

   Git：管理程式碼和檔案的版本，可以記錄修改歷史、建立分支，以及還原之前的內容。
   pyenv：管理及切換 Python 直譯器版本，例如 Python 3.10、3.12 和 3.13。
   venv：管理專案的獨立虛擬環境，使每個專案能使用自己的第三方套件及套件版本。

7. 本課程使用 GitHub Codespaces，
   為什麼通常不需要使用 pyenv？

   因為 GitHub Codespaces 會提供預先設定好的雲端開發環境，課程所需的 Python、工具及相關設定通常已經準備完成。所有使用相同 Codespaces 設定的人，也可以取得較一致的開發環境，因此通常不需要自行使用 pyenv 安裝或切換 Python。

   - 專案 B：Python 3.13

   先使用 pyenv 安裝兩個專案需要的 Python 版本：

#### 套件和虛擬環境

在大多數其他的程式語言之中， 一個套件( package ) 相當於一個程式庫(library)，都是由一組程式碼組成。Python 因為「一切需要的都已包含在內」而聞名， 大部分功能只需透過簡單的import陳述式就能取用。但如果您需要處理一些超過基本功能的事情， 例如建置華麗的使用者介面， 通常就需要安裝另一個套件來配合。

還好大多數第三方程式庫的安裝都相當簡單。程式庫的作者把它們打包成套件，「pip 套件管理器」可以用來安裝這些套件。

使用多個第三方套件需要一些技巧。有些套件需要先安裝其他套件來配合， 有些套件還可能與其他套件衝突，我們還可以根據需要安裝特定版本的套件。之前是否有提到電腦上的某些應用程式和作業系統元件依賴於特定的Python套件呢？這就是為什麼虛擬環境存在的原因。

虛擬環境就像沙盒， 您可以在其中只安裝特定於某個專案所需的Python套件，避免這些套件與其他專案（或您的系統）產生衝突的風險。您可以為每個專案建立不同的小沙盒， 只在其中安裝需要的套件， 讓一切都能保持整齊有序。實際上我們並沒有改變系統上已安裝的Python 套件， 因此可以避免破壞與您的專案無關但又很重要的事物。

#### 建立虛擬環境

每個虛擬環境都放在一個專屬的目錄中。通常我們會把這個資料夾命名為env或venv。對於每個專案， 我通常喜歡在專案資料夾內建立一個專屬的虛擬環境。Python提供了一個名為venv 的工具來完成這個工作。如果您使用Git或其他版本控制系統( version control system, VCS ) 來追蹤程
式碼的變更， 還有一個額外的設定步驟在後面會提到。

若想要在目前的工作目錄中建立名為venv 的虛擬環境， 請在命令列中執行以下指令：
```bash
python -m venv .venv
```
第一個venv是一個用來建立虛擬環境的指令，而第二個venv是指定虛擬環境的路徑。在上述的範例中，venv 只是個相對路徑，會在目前的工作目錄
中建立一個「.venv/ 」目錄。不過，也可以使用絕對路徑 並且可以任意取名。現在如果査看專案的工作目錄， 就會看到「venv/ 」資料夾已經被建立了。

若使用 Git 或 GitHub 進行版本控制，通常不會將 .venv 目錄加入版本控制，而是將其加入 .gitignore：
```bash
.venv/
```
這是因為虛擬環境中包含大量可重新建立的檔案，不需要上傳至 GitHub。相反地，通常會將套件資訊記錄在 requirements.txt 中。
```bash
pip freeze > requirements.txt
```

其他開發者便可利用：
```bash
pip install -r requirements.txt
```
重新建立相同的套件環境。

- 在 Linux、macOS 或 GitHub Codespaces 中，可使用下列指令啟用虛擬環境：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows 的命令提示字元（Command Prompt）中，請使用：
```bash
.venv\Scripts\activate.bat
```
- 在 Windows PowerShell 中，請使用：
```bash
.venv\Scripts\Activate.ps1
```
若 PowerShell 出現執行政策（Execution Policy）相關錯誤，可在 PowerShell 中執行：
```
Set-ExecutionPolicy RemoteSigned -Scope CurrentUser
```

啟用成功後，命令提示字元前方通常會出現虛擬環境名稱，例如：
```bash
(.venv) student@codespaces:~/project$
```
其中`(.venv)`表示目前正在使用名為 .venv 的虛擬環境。此時執行：`python --version`或`pip list`都將使用虛擬環境中的 Python 與套件，而不會影響系統中的其他 Python 專案。當工作完成後，可使用下列指令離開虛擬環境：`deactivate`。執行後，命令提示字元前方的`(.venv)`將會消失，表示已回到系統預設的 Python 環境。

本課程主要使用 GitHub Codespaces，因此通常只需要記住： 
```bash
python -m venv .venv 

source .venv/bin/activate 

deactivate
```

#### 課堂練習

請回答：

1 為什麼不建議將 .venv 目錄上傳到 GitHub？
.venv 裡面包含 Python 執行檔及已安裝的第三方套件，通常檔案很多、占用空間大，而且內容可能與作業系統、Python 版本及電腦路徑有關。其他人下載後不一定能直接使用，因此應將 .venv 加入 .gitignore，不要上傳到 GitHub。官方 Python Packaging Guide 也建議將虛擬環境目錄排除在版本控制系統之外。

2 為什麼建議將 requirements.txt 加入 Git 儲存庫？
requirements.txt 記錄專案所需要的套件及版本，例如：
Flask==3.1.2
requests==2.32.5
將此檔案加入 Git 儲存庫後，其他人取得專案時，就能知道需要安裝哪些套件，並可使用下列指令重建相近的執行環境：
pip install -r requirements.txt
``
因此，.venv 本身不需要分享，但建立 .venv 所需的套件清單應該分享。

3 requirements.txt 如何協助團隊合作？
requirements.txt 可以讓所有組員安裝相同的套件及版本，降低「在某位組員的電腦可以執行，但在另一位組員的電腦無法執行」的情況。

組員取得專案後，可以依序執行：
python -m venv .venv
啟用虛擬環境後，再執行：
pip install -r requirements.txt

4 如果兩位組員安裝了不同版本的 Flask，可能產生哪些問題？
不同版本的 Flask 可能具有不同的功能、函式介面、預設行為或相依套件，因此可能產生下列問題：
1.程式在一位組員的電腦可以執行，在另一位組員的電腦卻出錯。
2.新版 Flask 的功能在舊版中不存在，可能產生 ImportError 或 AttributeError。
3.舊功能可能在新版中被棄用、修改或移除。
4.測試結果可能不一致，增加除錯困難。
5.專案部署到伺服器後，可能因 Flask 版本不同而無法正常啟動。
6.Flask 所依賴的其他套件版本也可能不同，進一步造成相依性衝突。

因此，可以在 requirements.txt 中指定版本，例如：
Flask==3.1.2
所有組員再執行：
pip install -r requirements.txt
便能安裝相同版本，降低因套件版本不同造成的問題。不同專案或成員若使用不相容的套件版本，可能導致程式無法正常執行，而虛擬環境與需求檔可以協助隔離及重現套件環境。

請回答：

為什麼安裝一個套件時，可能會同時安裝許多其他套件？
因為一個套件通常不會包含所有功能，而是會使用其他套件所提供的功能。這些被需要的套件稱為相依套件（Dependencies）。

為什麼移除套件時需要注意相依性（Dependencies）？
因為準備移除的套件，可能同時被其他套件使用。如果移除共同需要的相依套件，其他套件可能會無法正常執行。

為什麼使用虛擬環境後，可以更放心地安裝與移除套件？
因為虛擬環境會為每個專案建立獨立的 Python 套件空間。在某個虛擬環境中安裝、升級或移除套件，通常只會影響該環境，不會影響其他專案或系統中的 Python 套件。

如果虛擬環境損壞或套件版本混亂，你會如何恢復開發環境？
可以刪除原本損壞的 .venv，再根據 requirements.txt 重新建立乾淨的虛擬環境。虛擬環境本來就應該可以透過套件需求檔重新建置。

如何判斷一個套件是否值得使用？
可以從以下幾個方面判斷：

來源是否可靠
 優先從 Python Package Index，也就是 PyPI，或套件官方網站取得安裝資訊。PyPI 是 Python 社群發布及取得套件的主要軟體儲存庫。

是否有清楚的官方文件
 良好的套件通常會提供安裝方法、範例、API 說明及版本更新資訊。

是否持續維護
 查看最近是否仍有版本更新、錯誤修正或開發活動。

使用人數與社群是否活躍
 可以參考下載量、GitHub 使用情況、Issue 回覆速度及社群討論，但人氣高不代表一定完全安全。

是否支援目前的 Python 版本
 套件可能只支援特定 Python 版本，安裝前應查看相容性說明。

授權條款是否適合專案
 如果是商業專案，應確認套件的開源授權是否允許商業使用與修改。

是否存在已知安全漏洞
 應查看安全公告，必要時使用安全掃描工具檢查相依套件。

功能是否真的符合需求
 不應只因套件熱門就安裝，也要考慮是否過於龐大、相依套件是否過多，以及有沒有更簡單的替代方案。

為什麼不應該隨意安裝來路不明的套件？
安裝 Python 套件不只是下載普通資料。套件在安裝或執行時，可能在電腦上執行程式碼，因此惡意套件可能造成：

竊取密碼、API 金鑰或個人資料
讀取或刪除電腦中的檔案
安裝惡意程式
將資料傳送到外部伺服器
控制電腦或消耗系統資源
透過相依套件引入更多安全漏洞
因名稱相似而誤裝仿冒套件

PyPI 將名稱仿冒、相依性混淆、資料外洩及惡意混淆程式碼等列為可能的惡意套件問題。

[!WARNING]

不要隨意使用 sudo pip

在 Linux 或 macOS 系統中，除非非常清楚自己在做什麼，否則不建議使用：

sudo pip install package_name 這個指令可能將套件安裝到系統 Python 環境中， 並覆蓋或干擾作業系統原本的套件設定。

輕則造成 Python 環境混亂， 重則可能影響系統工具正常運作。

對於大多數 Python 專案而言， 使用虛擬環境（venv）即可解決套件安裝與版本管理問題。

建議流程如下：

- python -m venv .venv
- source .venv/bin/activate
- pip install package_name

如此套件只會安裝在目前專案的虛擬環境中， 不會影響系統中的其他 Python 專案。

在本課程使用的 GitHub Codespaces 環境中， 原則上不需要使用 sudo pip。

#### 課堂練習

請回答：

1. 為什麼不建議使用 `sudo pip install`？
sudo pip install 會讓 pip 以最高系統權限安裝套件，可能直接修改系統層級的 Python 環境，造成以下問題：

覆蓋或修改作業系統需要的 Python 套件
與系統套件管理工具產生衝突
套件檔案變成由 root 擁有，日後一般使用者可能無法更新或移除
不同專案共用同一套套件，容易發生版本衝突
如果套件含有惡意程式碼，會以較高權限執行，增加安全風險

因此，安裝 Python 套件時通常應使用虛擬環境，而不是使用 sudo。Python Packaging User Guide 也建議使用虛擬環境來安裝第三方套件。

2. 虛擬環境（venv）如何降低套件安裝所帶來的風險？
虛擬環境會替專案建立獨立的 Python 套件空間。在虛擬環境中安裝、升級或移除套件時，通常只會影響目前專案，不會修改系統 Python，也不會影響其他專案。

3. 如果專案需要額外套件，應該安裝到哪裡？
應該安裝到該專案自己的虛擬環境，例如 .venv，而不是安裝到系統 Python。

4. 本課程使用 GitHub Codespaces，是否需要使用 `sudo pip`？為什麼？
通常不需要使用sudo pip。

GitHub Codespaces 的開發容器已經預先提供 Python、pip 及其他常用開發工具，專案套件可直接安裝到虛擬環境，或安裝到 Codespace 中可寫入的 Python 環境，不需要取得系統管理員權限。
