請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

已知下列資料：

```python
scores = [60, 70, 80]

settings = {
    "multiplier": 1.1,
    "bonus": 5
}
```

```python
result = analyze_scores(*scores, **settings)
```

## 四、總結

請根據本題，說明 `*`、`**` 與 `return` 在 Python 中的不同用途，並舉出至少一個程式碼例子，展示同一個符號在不同語法位置所代表的意義。

*、** 與 return 在 Python 中有不同的用途。

* 可以用來收集不固定數量的位置參數，也可以用於乘法運算。

** 可以用來收集不固定數量的關鍵字參數，也可以用於次方運算。

return 用來將函數計算後的結果回傳給呼叫端，使結果能夠被儲存或進一步處理。

例如：

def calculate(a, b):
    return a ** b

result = calculate(2, 3)

在這個例子中，** 表示次方運算，因此 2 ** 3 的結果為 8；return 則將計算結果回傳給變數 result。

```python

def calculate(a, b):
    return a ** b

result = calculate(2, 3)

```
在此例中，** 代表次方運算，2 ** 3 = 8。

return 將結果 8 回傳給呼叫端，因此 result 的值為 8。
